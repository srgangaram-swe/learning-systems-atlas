# Reinforcement learning tour

A thin client of the typed, transactional Atlas runner. Python 3.11–3.13; offline bundled/synthetic source; seed 42. No private data or production claim.


```python
from pathlib import Path
from tempfile import TemporaryDirectory

from IPython.display import Image, display

import learning_atlas
from learning_atlas.core.config import load_config
from learning_atlas.core.runner import run_experiment

root = Path(learning_atlas.__file__).resolve().parents[2]
config = load_config(root / "configs/reinforcement/q_learning.yaml")
workspace = TemporaryDirectory(prefix="atlas-tour-")
output = Path(workspace.name) / "run"
result = run_experiment(config, output)
print(result.selected_model)
print(result.metrics)
```

## Inspect every candidate and its limits

Use the recorded selection rule, units, baseline and held-out metrics. The underlying plotting API uses Seaborn; the notebook does not recreate estimator logic.


```python
for candidate in result.candidates:
    print(candidate.name, candidate.metrics)
for relative in result.artifacts.values():
    if relative.endswith(".png"):
        display(Image(filename=str(output / relative)))
print("\n".join(result.notes))
```
